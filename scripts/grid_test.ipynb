```python
from multigrid.core.world_object import Wall
from multigrid.utils.obs import get_view_exts
def get_agent_obs_grid(
    grid: Grid,
    agent: Agent,
    agent_view_size: int,
) -> Grid:
    agent_dir = np.array([agent.state.dir.value])
    agent_pos = np.array([agent.state.pos])

    top_left = get_view_exts(agent_dir, agent_pos, agent_view_size)
    topX, topY = top_left[0, 0], top_left[0, 1]

    num_left_rotations = (agent_dir[0] + 1) % 4
    obs_grid = Grid(agent_view_size, agent_view_size)

    for i in range(agent_view_size):
        for j in range(agent_view_size):
            x, y = topX + i, topY + j

            # Apply same rotation mapping as gen_obs_grid
            if num_left_rotations == 0:
                i_rot, j_rot = i, j
            elif num_left_rotations == 1:
                i_rot, j_rot = j, agent_view_size - i - 1
            elif num_left_rotations == 2:
                i_rot, j_rot = agent_view_size - i - 1, agent_view_size - j - 1
            elif num_left_rotations == 3:
                i_rot, j_rot = agent_view_size - j - 1, i

            if 0 <= x < grid.width and 0 <= y < grid.height:
                obs_grid.set(i_rot, j_rot, grid.get(x, y))
            else:
                obs_grid.set(i_rot, j_rot, Wall())

    return obs_grid
```

```python
from multigrid.utils.obs import gen_obs_grid
from multigrid.core.constants import TILE_PIXELS
from numpy.typing import NDArray as ndarray
from multigrid.core.agent import AgentState
def get_all_agent_obs_images(
    grid: Grid,
    agents: list[Agent],
    agent_view_size: int,
    tile_size: int = 6,
) -> ndarray:
    num_agents = len(agents)
    obs_images = np.zeros(
        (num_agents, agent_view_size * tile_size, agent_view_size * tile_size, 3),
        dtype=np.uint8
    )

    for agent_idx, agent in enumerate(agents):
        obs_grid = get_agent_obs_grid(grid, agent, agent_view_size)
        
        # Place ego agent at bottom center facing up
        ego_agent = Agent(index=agent_idx)
        ego_agent.state.pos = np.array([agent_view_size // 2, agent_view_size - 1])
        ego_agent.state.dir = Direction.up

        obs_images[agent_idx] = obs_grid.render(tile_size, agents=[ego_agent])
        # print(obs_grid.render(tile_size, agents=[ego_agent]).shape)

    return obs_images
```

```python
from multigrid.utils.obs import gen_obs_grid
from multigrid.core.constants import TILE_PIXELS
from numpy.typing import NDArray as ndarray
def get_all_agent_obs_images(
    grid_state: ndarray,
    agent_state: ndarray,
    agent_view_size: int,
    agents: list[Agent],
    tile_size: int = TILE_PIXELS,
) -> ndarray:
    obs_grid = gen_obs_grid(grid_state, agent_state, agent_view_size)  # already vectorized
    num_agents = len(agents)
    obs_images = np.zeros((num_agents, agent_view_size * tile_size, agent_view_size * tile_size, 3), dtype=np.uint8)

    for agent_idx in range(num_agents):
        grid, _ = Grid.decode(obs_grid[agent_idx])
        obs_images[agent_idx] = grid.render(tile_size, agents=(agents[agent_idx],))

    return obs_images
```

```python
width, height = 15, 15
grid = Grid(width, height)
grid.wall_rect(0, 0, width, height)

from multigrid.core.world_object import Goal
goal = Goal()
grid.set(3, 3, goal)
goal.init_pos = (3, 3)
goal.cur_pos = (3, 3)

from multigrid.core.agent import Agent
import numpy as np
```

```python
from multigrid.core.agent import AgentState
agents = 1
mission_space = None
agent_view_size = 7
see_through_walls = False
num_agents = agents
agent_states = AgentState(agents) # joint agent state (vectorized)
self_agents: list[Agent] = []
for i in range(agents):
    agent = Agent(
        index=i,
        # mission_space=mission_space,
        view_size=agent_view_size,
        see_through_walls=see_through_walls,
    )
    agent.state = agent_states[i]
    self_agents.append(agent)
```

```python
self_agents[0].state.pos = np.array([3, 2])
self_agents[0].state.dir = Direction.right
```

```python
render_grid(grid, self_agents)
```
