One id per warehouse size, as RWARE's presets; the number of robots, the difficulty, the observations
and the rest are arguments, e.g. ``make('RWARE-Small-v0', agents=4, difficulty='hard', observations=['vector'])``.

```python
#| export
from stable_marl.envs.registry import register


def _make_rware(**kwargs):
    from stable_marl.envs.rware.env import RwareEnv   # imported on first use (networkx)
    return RwareEnv(**kwargs)


CONFIGURATIONS = {f'RWARE-{size.capitalize()}-v0': {'size': size} for size in ('tiny', 'small', 'medium', 'large')}
for name, config in CONFIGURATIONS.items():
    register(name, _make_rware, config, family='rware')

_all_ = ['CONFIGURATIONS']
```

```python
#| hide
from stable_marl.envs import make, list_envs
assert list_envs('rware') == ['RWARE-Large-v0', 'RWARE-Medium-v0', 'RWARE-Small-v0', 'RWARE-Tiny-v0']
env = make('RWARE-Small-v0', agents=4, difficulty='hard')
assert (env.height, env.width, env.num_agents, env.warehouse.request_queue_size) == (20, 10, 4, 2)
```
