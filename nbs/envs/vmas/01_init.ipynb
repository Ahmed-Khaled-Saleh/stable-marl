One id per VMAS scenario (VMAS 1.5's 21 scenarios and 9 MPE tasks), e.g. `VMAS-Navigation-v0`,
`VMAS-SimpleSpread-v0`; the scenario's arguments, `continuous_actions` and `max_steps` are arguments,
e.g. ``make('VMAS-Transport-v0', n_agents=3, continuous_actions=False)``. Creating them needs the `vmas`
package (`pip install 'stable-marl[vmas]'`).

```python
#| export
from stable_marl.envs.registry import register

SCENARIOS = ('balance', 'ball_passage', 'ball_trajectory', 'buzz_wire', 'discovery', 'dispersion', 'dropout', 'flocking',
             'football', 'give_way', 'joint_passage', 'joint_passage_size', 'multi_give_way', 'navigation', 'passage',
             'reverse_transport', 'road_traffic', 'sampling', 'transport', 'wheel', 'wind_flocking')
MPE_SCENARIOS = ('simple', 'simple_adversary', 'simple_crypto', 'simple_push', 'simple_reference', 'simple_speaker_listener',
                 'simple_spread', 'simple_tag', 'simple_world_comm')


def env_id(scenario: str) -> str:
    "The stable-marl id of a VMAS scenario: 'simple_spread' -> 'VMAS-SimpleSpread-v0'."
    return f"VMAS-{''.join(p.capitalize() for p in scenario.split('_'))}-v0"


def _make_vmas(**kwargs):
    try:
        from stable_marl.envs.vmas.env import VmasEnv
        return VmasEnv(**kwargs)
    except ModuleNotFoundError as e:
        if e.name == 'vmas':
            raise ModuleNotFoundError("The VMAS environments need the vmas package: pip install 'stable-marl[vmas]'") from e
        raise


for scenario in (*SCENARIOS, *MPE_SCENARIOS):
    register(env_id(scenario), _make_vmas, {'scenario': scenario}, family='vmas')

_all_ = ['SCENARIOS', 'MPE_SCENARIOS', 'env_id']
```

```python
#| hide
import vmas
from stable_marl.envs import make, list_envs
assert set(SCENARIOS) == set(vmas.scenarios) and set(MPE_SCENARIOS) == set(vmas.mpe_scenarios)   # VMAS's list
assert len(list_envs('vmas')) == 30 and env_id('joint_passage_size') == 'VMAS-JointPassageSize-v0'
env = make('VMAS-Transport-v0', n_agents=3, continuous_actions=False)
assert env.num_agents == 3 and env.action_space[0].n == 9
```
