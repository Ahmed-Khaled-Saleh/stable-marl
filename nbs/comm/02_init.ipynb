`stable_marl.comm` connects [comm-core](https://github.com/Ahmed-Khaled-Saleh/comm-core) to stable-marl's agents:
`AgentNetwork` (a comm-core network over the agents of one env, for methods that exchange things themselves, e.g.
world-model latents at planning time) and `CommunicationWrapper` / `communication` (communication as part of the
environment). Needs comm-core (`pip install 'stable-marl[comm]'`), imported when a network is created.
