```python
#| export
from stable_marl.envs.multigrid.base import MultiGridEnv
from stable_marl.envs.multigrid.blockedunlockpickup import BlockedUnlockPickupEnv
from stable_marl.envs.multigrid.empty import EmptyEnv
from stable_marl.envs.multigrid.locked_hallway import LockedHallwayEnv
from stable_marl.envs.multigrid.playground import PlaygroundEnv
from stable_marl.envs.multigrid.redbluedoors import RedBlueDoorsEnv
from stable_marl.envs.multigrid.findgoal import FindGoalEnv
from stable_marl.envs.multigrid.expert import GoToGoalPolicy

_all_ = ['MultiGridEnv', 'BlockedUnlockPickupEnv', 'EmptyEnv', 'LockedHallwayEnv',
         'PlaygroundEnv', 'RedBlueDoorsEnv', 'FindGoalEnv', 'GoToGoalPolicy']
```

```python
#| export
CONFIGURATIONS = {
    'MultiGrid-BlockedUnlockPickup-v0': (BlockedUnlockPickupEnv, {}),
    'MultiGrid-Empty-5x5-v0': (EmptyEnv, {'size': 5}),
    'MultiGrid-Empty-Random-5x5-v0': (EmptyEnv, {'size': 5, 'agent_start_pos': None}),
    'MultiGrid-Empty-6x6-v0': (EmptyEnv, {'size': 6}),
    'MultiGrid-Empty-Random-6x6-v0': (EmptyEnv, {'size': 6, 'agent_start_pos': None}),
    'MultiGrid-Empty-8x8-v0': (EmptyEnv, {}),
    'MultiGrid-Empty-16x16-v0': (EmptyEnv, {'size': 16}),
    'MultiGrid-LockedHallway-2Rooms-v0': (LockedHallwayEnv, {'num_rooms': 2}),
    'MultiGrid-LockedHallway-4Rooms-v0': (LockedHallwayEnv, {'num_rooms': 4}),
    'MultiGrid-LockedHallway-6Rooms-v0': (LockedHallwayEnv, {'num_rooms': 6}),
    'MultiGrid-Playground-v0': (PlaygroundEnv, {}),
    'MultiGrid-RedBlueDoors-6x6-v0': (RedBlueDoorsEnv, {'size': 6}),
    'MultiGrid-RedBlueDoors-8x8-v0': (RedBlueDoorsEnv, {'size': 8}),
    'MultiGrid-FindGoal-15x15-v0': (FindGoalEnv, {'size': 15})
}
```

```python
#| export
# Register environments (in stable-marl and gymnasium)
from stable_marl.envs.registry import register
for name, (env_cls, config) in CONFIGURATIONS.items():
    register(name, env_cls, config, family='multigrid')
```
