One id per multi-arm RoboFactory task, e.g. `RoboFactory-LiftBarrier-v0`, `RoboFactory-TakePhoto-v0`
(RoboFactory's three single-arm tasks are left out); the scene, observations, camera size and step limit
are arguments, e.g. `make('RoboFactory-PassShoe-v0', scene='robocasa', observations=('vector',))`.
Creating them needs ManiSkill (`pip install 'stable-marl[robofactory]'`).

```python
#| export
from stable_marl.envs.registry import register
from stable_marl.envs.robofactory.env import TASKS, SCENES, RoboFactoryEnv, asset_dir, download_assets
from stable_marl.envs.robofactory.expert import RoboFactoryExpert

def env_id(task: str) -> str:
    "The stable-marl id of a RoboFactory task: 'LiftBarrier' -> 'RoboFactory-LiftBarrier-v0'."
    return f'RoboFactory-{task}-v0'

def _make_robofactory(**kwargs):
    try:
        return RoboFactoryEnv(**kwargs)
    except ModuleNotFoundError as e:
        if e.name in ('mani_skill', 'sapien'):
            raise ModuleNotFoundError("The RoboFactory environments need ManiSkill: pip install 'stable-marl[robofactory]'") from e
        raise

for task in TASKS:
    register(env_id(task), _make_robofactory, {'task': task}, family='robofactory')

_all_ = ['TASKS', 'SCENES', 'RoboFactoryEnv', 'RoboFactoryExpert', 'asset_dir', 'download_assets', 'env_id']
```
