stable-worldmodel's wrappers live in `stable_worldmodel.wrapper` (`.default`, `.visual`); ours in
`stable_marl.wrappers`. `stable_marl.wrapper` re-exports the same objects under stable-worldmodel's
names, so code written for it only needs its package name changed. The visual wrappers stay lazy
(they import OpenCV).

```python
#| export
import importlib

from stable_marl.wrappers.default import (AddPixelsWrapper, EnsureGoalInfoWrapper, EnsureImageShape,
                                          EnsureInfoKeysWrapper, EverythingToInfoWrapper, MapKeysWrapper, MegaWrapper,
                                          ResizeGoalWrapper, TransformInfoWrapper)


def __getattr__(name: str):
    "Everything else (the visual wrappers, loaded lazily) from `stable_marl.wrappers`."
    if name.startswith('__'):
        raise AttributeError(name)
    return getattr(importlib.import_module('stable_marl.wrappers'), name)

_all_ = ['AddPixelsWrapper', 'EnsureGoalInfoWrapper', 'EnsureImageShape', 'EnsureInfoKeysWrapper',
         'EverythingToInfoWrapper', 'MapKeysWrapper', 'MegaWrapper', 'ResizeGoalWrapper', 'TransformInfoWrapper']
```

```python
#| hide
import subprocess, sys
code = """
import sys
import stable_marl.wrapper as w, stable_marl.wrappers as ws
from stable_marl.wrapper.default import MegaWrapper, MapKeysWrapper
assert MegaWrapper is ws.MegaWrapper and w.MapKeysWrapper is MapKeysWrapper
assert 'cv2' not in sys.modules
from stable_marl.wrapper.visual import GrayscaleWrapper
assert GrayscaleWrapper is ws.GrayscaleWrapper and w.NoiseWrapper is ws.NoiseWrapper
"""
subprocess.run([sys.executable, '-W', 'ignore', '-c', code], check=True)
```
